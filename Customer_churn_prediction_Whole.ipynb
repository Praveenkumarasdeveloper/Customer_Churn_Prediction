# Task 2 — Clean the Dataset.

```python
import pandas as pd

def Cleaned_dataset():
    # 1. Load churnguard_data.csv.
    
    Internet_Bill = pd.read_csv("churnguard_raw_data.csv")
    Internet_Bill_Clean = Internet_Bill.copy()
    
    # 2. Drop the customerID column.
    
    Internet_Bill_Clean = Internet_Bill_Clean.drop(columns = "customerID")
    
    # 3. Remove duplicate rows.
    
    Internet_Bill_Clean = Internet_Bill_Clean.drop_duplicates()
    
    # 4. Strip whitespace from gender and PaymentMethod using .str.strip()
    
    Internet_Bill_Clean["gender"].value_counts()  # sum() = 1030
    Internet_Bill_Clean["gender"] = Internet_Bill_Clean["gender"].str.strip()
    
    Internet_Bill_Clean["PaymentMethod"] = Internet_Bill_Clean["PaymentMethod"].str.strip()
    Internet_Bill_Clean["PaymentMethod"].value_counts() # sum() = 1030
    
    # 5. Standardise casing — convert Churn, PhoneService, and PaperlessBilling to title case using .str.strip().str.title()
    
    Internet_Bill_Clean["PhoneService"] = Internet_Bill_Clean["PhoneService"].str.strip().str.title()
    Internet_Bill_Clean["PhoneService"].value_counts() # sum() = 1030 
    
    Internet_Bill_Clean["PaperlessBilling"] = Internet_Bill_Clean["PaperlessBilling"].str.strip().str.title()
    Internet_Bill_Clean["PaperlessBilling"].value_counts() # sum() = 1030
    
    Internet_Bill_Clean["Churn"] = Internet_Bill_Clean["Churn"].str.strip().str.title()
    Internet_Bill_Clean["Churn"].value_counts() # sum() = 1030
    
    # 6. Fix Contract — map all variations to one of three valid values.
    
    Internet_Bill_Clean["Contract"] = Internet_Bill_Clean["Contract"].str.strip().replace({"month-to-month":"Month-to-month",
                                          "month to month":"Month-to-month",
                                          "Monthly":"Month-to-month",
                                            "monthly":"Month-to-month",
                                          "1 year":"One year",
                                          "one year":"One year",
                                          "2 year":"Two year",
                                          "two year":"Two year"})
    Internet_Bill_Clean["Contract"].value_counts() # sum() = 1030
    
    # 7. Fix InternetService — map all variations to one of three valid values.
    
    Internet_Bill_Clean["InternetService"] = Internet_Bill_Clean["InternetService"].str.strip().replace({"dsl":"DSL",
                                                                      "DSl":"DSL",
                                                                      "Dsl":"DSL",
                                                                      "Fibre optic":"Fiber optic",
                                                                      "fiber optic":"Fiber optic",
                                                                      "FiberOptic":"Fiber optic",
                                                                      "NO":"No",
                                                                      "no":"No",
                                                                      "fibre optic":"Fiber optic",
                                                                      "fiberoptic":"Fiber optic",
                                                                      })
   
    
    # 8. Fix TotalCharges — convert to numeric using pd.to_numeric(..., errors='coerce') so junk becomes NaN
    
    Internet_Bill_Clean["TotalCharges"].value_counts() # .sum()
    Internet_Bill_Clean["TotalCharges"] = pd.to_numeric(Internet_Bill_Clean["TotalCharges"],errors='coerce')
    
    # 9. Remove rows where tenure is zero or negative.
    
    Internet_Bill_Clean = Internet_Bill_Clean[Internet_Bill_Clean["tenure"] > 0]
    
    # 10. Remove rows where MonthlyCharges is less than 10 or greater than 200.
    
    Internet_Bill_Clean = Internet_Bill_Clean[(Internet_Bill_Clean["MonthlyCharges"] >= 10.00) & (Internet_Bill_Clean["MonthlyCharges"] <= 200.00)]
    
    # 11. Fill missing values.

    Internet_Bill_Clean["MonthlyCharges"].isna().sum() # There are no missing values in MonthlyCharges.
    
    Internet_Bill_Clean["tenure"].isna().sum()  # There are no missing values in tenure.
    
    Mean_TotalCharges = Internet_Bill_Clean["TotalCharges"].mean() 
    Internet_Bill_Clean["TotalCharges"] = Internet_Bill_Clean["TotalCharges"].fillna(Mean_TotalCharges)
    
    # 12. Print the shape of the cleaned DataFrame.
    
    print("------Shape of the cleaned DataFrame------\n")
    
    Internet_Bill_Cleaned_shape = Internet_Bill_Clean.shape
    
    print(Internet_Bill_Cleaned_shape,"\n")
    
    # 13. Print missing value counts to confirm all issues are resolved.
    
    print("------Missing value counts to confirm all issues are resolved------\n")
    
    Internet_Bill_Cleaned_Missing_Value = Internet_Bill_Clean.isna().sum()
    
    print(Internet_Bill_Cleaned_Missing_Value)

Cleaned_dataset()
```

# Task 3 — Train a Classification Model.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report


def Train_data():
    #  For this step load the task 2 data set here.
    # 1. Train a Classification Model.
    
    Internet_Bill_Clean = pd.read_csv("Internet_Bill_Clean.csv")
    Internet_Bill_Train = Internet_Bill_Clean.copy()
    
    # 2. Encode the target column Churn: Yes → 1, No → 0.
    
    Internet_Bill_Train["Churn"]= Internet_Bill_Train["Churn"].replace({"Yes":1,
                                                                        "No":0})
    
    # 3. Encode categorical columns using pd.get_dummies() with drop_first=True.
    
    Internet_Bill_Train_dummies = pd.get_dummies(Internet_Bill_Train,columns = ["gender",
                                                    "PhoneService",
                                                    "InternetService",
                                                    "Contract",
                                                    "PaperlessBilling",
                                                    "PaymentMethod"])
    Internet_Bill_Train_dummies
    
    # 4. Separate the data into.
    # 4.1 _ X — all columns except Churn.
    # 4.2 _ y — the Churn column.
    
    X = Internet_Bill_Train_dummies.drop(columns = ["Churn"])
    Y = Internet_Bill_Train_dummies["Churn"]
    
    # 5. Split into train and test sets — 80% train, 20% test, random_state=42.
    
    
    X_train, X_test, Y_train, Y_test = train_test_split(X, 
                                                        Y, 
                                                        test_size = 0.2, 
                                                        random_state = 42)
    
    # 6. Train a LogisticRegression model with max_iter=1000.
    
    
    model = LogisticRegression(max_iter=1000)
    model.fit(X_train,Y_train)
    
    # 7. Print the accuracy score on the test set.
    
    
    Y_pred = model.predict(X_test)
    accuracy = accuracy_score(Y_test,Y_pred)
    
    print("1. ------Accuracy_Score------\n")
    
    print(f"Accuracy_score_of_this_model_is {accuracy}\n")

    print(f"Accuracy_score_of_this_model_in_percentage {accuracy*100} % \n")
    
    # 8. Print the classification report using classification_report with target_names=['Stay', 'Churn']
    
    
    
    Classification_report = classification_report(Y_test,
                                Y_pred,
                                target_names=["Stay","Churn"])
    
    print("2. ------Classification_Report------\n")
    
    print(f"Classification_report_of_this_model_is \n {Classification_report}")

Train_data()
```

# Task 4 — Build a Prediction Script


```python
import pandas as pd 
from sklearn.linear_model import LogisticRegression


def prediction_script():
    # 1. Load and clean churnguard_data.csv (apply all Task 2 cleaning steps)
    
    Internet_Bill_Predict = pd.read_csv("Internet_Bill_Clean.csv")
    Internet_Bill_Predict_Script = Internet_Bill_Predict.copy()
    
    # 2. Encode Churn as Yes → 1, No → 0
    
    Internet_Bill_Predict_Script["Churn"] = Internet_Bill_Predict_Script["Churn"].replace({"Yes":1,
                                          "No":0})
    
    # 3. Retrain a LogisticRegression model on the full cleaned dataset — no train/test split needed here
    
    Internet_Bill_Predict_Script["Contract"]= Internet_Bill_Predict_Script["Contract"].replace({"Month-to-month":0,
                                                     "One year":1,
                                                     "Two year":2})
    
    feature = ["tenure","MonthlyCharges","TotalCharges","SeniorCitizen","Contract"]
    
    X = Internet_Bill_Predict_Script[feature]
    
    Y = Internet_Bill_Predict_Script["Churn"]
    
    
    
    model = LogisticRegression(max_iter = 1000)
    
    model.fit(X,Y)
    
    # 4. Collect the following inputs from the user using input():
    
    tenure = int(input("Enter tenure (months) : "))
    MonthlyCharges = float(input("Enter Monthly Charges : "))
    TotalCharges = float(input("Enter Total Charges : "))
    SeniorCitizen = int(input("Senior Citizen? (1 = Yes, 0 = No) : "))
    Contract = int(input("Contract type (0 = Month-to-month, 1 = One year, 2 = Two year):"))
    
    
    print("\n------Customer Details------")
    print("Tenure :",tenure)
    print("Monthlt Charges :",MonthlyCharges)
    print("TotalCharges :",TotalCharges)
    print("SeniorCitizen :",SeniorCitizen)
    print("Contract :",Contract)
    
    # 5. Use the trained model to predict churn for the entered values.
    
    new_customer = pd.DataFrame([{
                                    "tenure": tenure,
                                    "MonthlyCharges": MonthlyCharges,
                                    "TotalCharges": TotalCharges,
                                    "SeniorCitizen": SeniorCitizen,
                                    "Contract": Contract
                                   }])
    
    # prediction = 
    prediction = model.predict(new_customer)[0]
    
    if prediction == 1:
        print("Prediction : Customer is likely to CHURN.")
    
    else:
        print("Prediction : Customer is likely to STAY")

prediction_script()
```
