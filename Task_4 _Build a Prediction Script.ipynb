# 4.7 _ How each feature affects churn

**1. tenure — How long the customer has stayed.**

Example: 2 months vs. 48 months.

- Short tenure: the customer may be more likely to leave.
- Long tenure: the customer may be more likely to stay.

Why? Long-term customers may be more comfortable with the service. This is a tendency, not a rule.

**2. MonthlyCharges — Monthly bill amount.**

Example: ₹500 vs. ₹1,500.

- Higher charges may increase the chance of churn if customers feel the service is too expensive.
- Lower charges may make the service more affordable.

**3. TotalCharges — Total amount charged so far.**

Example: 513 vs. 5,000.

- A higher total may simply mean the customer has used the service longer or paid more each month.
- A lower total may indicate a newer customer or lower charges.
Important: TotalCharges is related to both tenure and MonthlyCharges, so interpret it alongside those features.

**4. SeniorCitizen — Whether the customer is a senior citizen.**
   
    - 0 = No
    - 1 = Yes
      
Senior-citizen status may be associated with different churn patterns in the dataset. The model learns this relationship from historical examples; it does not mean every senior citizen behaves the same way.

**5. Contract — Type of contract.**
 
Your code uses these numerical values:
Value	Contract
0	Month-to-month
1	One year
2	Two year

- Month-to-month customers can usually leave more easily.
- One-year and two-year contracts may be associated with lower churn because customers have longer commitments.

```python
import pandas as pd 

def prediction_script():
    # 1. Load and clean churnguard_data.csv (apply all Task 2 cleaning steps)
    
    Internet_Bill_Predict = pd.read_csv("Cleaned_data/Internet_Bill_Clean.csv")
    Internet_Bill_Predict_Script = Internet_Bill_Predict.copy()
    
    # 2. Encode Churn as Yes → 1, No → 0
    
    Internet_Bill_Predict_Script["Churn"] = Internet_Bill_Predict_Script["Churn"].replace({"Yes":1,
                                          "No":0})
    
    # 3. Retrain a LogisticRegression model on the full cleaned dataset — no train/test split needed here
    
    Internet_Bill_Predict_Script["Contract"]= Internet_Bill_Predict_Script["Contract"].replace({"Month-to-month":0,
                                                     "One year":1,
                                                     "Two year":2})
    
    feature = ["tenure","MonthlyCharges","TotalCharges","SeniorCitizen","Contract"]
    
    X = Internet_Bill_Predict_Script[feature]
    
    Y = Internet_Bill_Predict_Script["Churn"]
    
    from sklearn.linear_model import LogisticRegression
    
    model = LogisticRegression(max_iter = 1000)
    
    model.fit(X,Y)
    
    # 4. Collect the following inputs from the user using input():
    
    tenure = int(input("Enter tenure (months) : "))
    MonthlyCharges = float(input("Enter Monthly Charges : "))
    TotalCharges = float(input("Enter Total Charges : "))
    SeniorCitizen = int(input("Senior Citizen? (1 = Yes, 0 = No) : "))
    Contract = int(input("Contract type (0 = Month-to-month, 1 = One year, 2 = Two year):"))
    
    
    print("\n------Customer Details------")
    print("Tenure :",tenure)
    print("Monthlt Charges :",MonthlyCharges)
    print("TotalCharges :",TotalCharges)
    print("SeniorCitizen :",SeniorCitizen)
    print("Contract :",Contract)
    
    # 5. Use the trained model to predict churn for the entered values.
    
    new_customer = pd.DataFrame([{
                                    "tenure": tenure,
                                    "MonthlyCharges": MonthlyCharges,
                                    "TotalCharges": TotalCharges,
                                    "SeniorCitizen": SeniorCitizen,
                                    "Contract": Contract
                                   }])
    
    # prediction = 
    prediction = model.predict(new_customer)[0]
    
    if prediction == 1:
        print("Prediction : Customer is likely to CHURN.")
    
    else:
        print("Prediction : Customer is likely to STAY")

prediction_script()
```
